# Análisis de Datos de Transporte (GTFS)

Proyecto de análisis de datos de transporte listo para **Google Colab**.

**Pregunta:** ¿cómo analizar horarios y rutas de transporte público?

**Método:** Lectura de archivos GTFS

**Stack:** pandas, matplotlib

Datos de ejemplo (estructura GTFS).

## 0. Setup

```python
import pandas as pd
import matplotlib.pyplot as plt
print("OK")
```

## 1. Datos de ejemplo (estructura GTFS)

```python
# Ejemplo simplificado de stops y stop_times
stops = pd.DataFrame({
    "stop_id": [1, 2, 3, 4],
    "stop_name": ["Terminal", "Centro", "Hospital", "Universidad"]
})

stop_times = pd.DataFrame({
    "trip_id": [1, 1, 1, 2, 2],
    "stop_id": [1, 2, 3, 1, 4],
    "arrival_time": ["08:00:00", "08:15:00", "08:30:00", "09:00:00", "09:20:00"]
})
print(stops)
print(stop_times)
```

## 2. Unir datos

```python
merged = stop_times.merge(stops, on="stop_id")
print(merged)
```

## 3. Análisis simple

```python
print("Paradas más frecuentes:")
print(merged["stop_name"].value_counts())
```

## 4. Visualización

```python
merged["stop_name"].value_counts().plot(kind="bar")
plt.title("Frecuencia de paradas")
plt.ylabel("Cantidad")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

## 5. Conclusiones

- GTFS es el estándar para datos de transporte público.
- Se puede extender a análisis de rutas, frecuencias y visualización en mapas.
- Ideal para proyectos de movilidad urbana.
